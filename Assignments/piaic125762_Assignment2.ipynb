## Question:2

###  How to stack two arrays vertically?

#### Desired Output::

```python
a = np.arange(10).reshape(2, 5)
b = np.ones((2, 5), dtype=int)
np.vstack((a, b))
```

## Question:3

### How to stack two arrays horizontally?

#### Desired Output::

```python
np.hstack((a, b))
```

## Question:14

### Find the 5th and 95th percentile of an array?

```python
arr11 = np.arange(50)
print("5th percentile of arr : ", 
       np.percentile(arr11, 5))
print("95th percentile of arr : ", 
       np.percentile(arr11, 95))
```

## Question:15

### How to find if a given array has any null values?

```python
arr12 = np.arange(25).reshape(5, 5)
arr13 = np.array([[1, 2, 3, 4], [np.nan, 2, 2, 2]])
print("Has null values:", np.isnan(arr12).any())
print("Has null values:", np.isnan(arr13).any())
```
